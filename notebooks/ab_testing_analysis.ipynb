```python

import pandas as pd
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

print("Libraries imported successfully!")
```

```python
# Number of retained users
successes = [20034, 20119]

# Total users in each group
observations = [44700, 45489]

# Two-proportion Z-test
z_stat, p_value = proportions_ztest(
    count=successes,
    nobs=observations
)

print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.6f}")

alpha = 0.05

if p_value < alpha:
    print("Statistically significant difference")
else:
    print("No statistically significant difference")
```

```python
# Retention Day 7

successes_d7 = [8502, 8279]
observations = [44700, 45489]

z_stat_d7, p_value_d7 = proportions_ztest(
    count=successes_d7,
    nobs=observations
)

print(f"Z-statistic D7: {z_stat_d7:.4f}")
print(f"P-value D7: {p_value_d7:.6f}")

alpha = 0.05

if p_value_d7 < alpha:
    print("Statistically significant difference")
else:
    print("No statistically significant difference")
```

```python

import numpy as np
from statsmodels.stats.proportion import confint_proportions_2indep

# Sample sizes
n_control = 44700
n_treatment = 45489

# Retained users
metrics = {
    "Retention D1": (20034, 20119),
    "Retention D7": (8502, 8279)
}

for metric, (control, treatment) in metrics.items():

    p_control = control / n_control
    p_treatment = treatment / n_treatment

    # Treatment minus Control
    difference = p_treatment - p_control

    # 95% confidence interval (Newcombe method)
    ci_low, ci_high = confint_proportions_2indep(
        count1=treatment,
        nobs1=n_treatment,
        count2=control,
        nobs2=n_control,
        compare="diff",
        method="newcomb",
        alpha=0.05
    )

    print(f"\n{metric}")
    print(f"Difference: {difference * 100:.3f} pp")
    print(f"95% CI: [{ci_low * 100:.3f}, {ci_high * 100:.3f}] pp")

    if ci_low > 0 or ci_high < 0:
        print("95% CI excludes zero")
    else:
        print("95% CI includes zero")
```

```python

from scipy.stats import chisquare

# Observed users
observed = [44700, 45489]

# Expected 50/50 allocation
total_users = sum(observed)
expected = [total_users / 2] * 2

# Sample Ratio Mismatch test
chi2_stat, p_value_srm = chisquare(
    f_obs=observed,
    f_exp=expected
)

print(f"Chi-square statistic: {chi2_stat:.4f}")
print(f"SRM p-value: {p_value_srm:.6f}")

# Common SRM threshold
alpha_srm = 0.001

if p_value_srm < alpha_srm:
    print("Potential Sample Ratio Mismatch detected")
else:
    print("No statistically significant SRM detected")
```

### Sample Ratio Mismatch (SRM) Validation

A Chi-square goodness-of-fit test was conducted to evaluate whether the observed experiment allocation differed from the assumed 50/50 split.

**Results:**
- Control group: 44,700 users
- Treatment group: 45,489 users
- Chi-square statistic: 6.9024
- P-value: 0.008608
- SRM significance threshold: 0.001

**Interpretation:**

No statistically significant sample ratio mismatch was detected at the predefined threshold of 0.001.

However, the p-value is below 0.05, indicating some evidence of an allocation imbalance under the assumed 50/50 design.

Further validation of the experiment assignment mechanism would be recommended before making a final rollout decision.
